```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.metrics import mean_absolute_error

X_train,X_test,Y_train,Y_test=train_test_split(data[:,1:2],data[:,0])

param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5, 10],
    'max_features': ['sqrt', 'log2', None]
}

grid_search = GridSearchCV(
    estimator=RandomForestRegressor(random_state=42),
    param_grid=param_grid,
    scoring='neg_mean_absolute_error', 
    cv=5,
    n_jobs=-1,
    verbose=2
)

grid_search.fit(X_train,Y_train)
model=grid_search.best_estimator_

predictions=model.predict(X_test)

print(np.percentile(np.abs(Y_test-predictions),95))
print(mean_absolute_error(Y_test,predictions))
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

X_train,X_test,Y_train,Y_test=train_test_split(data[:,1],data[:,0:1])

# 1. Transform the Target (Distance) to Log Scale
# Physics expects: RSSI = P0 - 10 * n * log10(d)
log_Y_train = np.log10(Y_train).reshape(-1, 1)
log_Y_test = np.log10(Y_test).reshape(-1, 1)

# 2. Fit the Linear Model
# We swap roles: we find the relationship between log10(dist) and RSSI
physics_model = LinearRegression()
physics_model.fit(log_Y_train, X_train) # Features = log(dist), Target = RSSI

# 3. Extract Physics Parameters
p0 = physics_model.intercept_
# Slope m = -10n -> n = -m / 10
n_exponent = -physics_model.coef_[0] / 10

print(f"Calculated P0 (RSSI at 1m): {p0:.2f} dBm")
print(f"Calculated Path Loss Exponent (n): {n_exponent:.2f}")

# 4. Predict Distance from RSSI
# Rearranged formula: d = 10 ^ ((P0 - RSSI) / (10 * n))
def predict_distance(rssi, p0, n):
    return 10**((p0 - rssi) / (10 * n))

# Test the model on the test set
Y_pred_dist = predict_distance(X_test, p0, n_exponent)

# 5. Visualization
plt.figure(figsize=(10, 6))
plt.scatter(X_test, Y_test, color='blue', alpha=0.3, label='Actual Data')
sorted_idx = np.argsort(X_test)
plt.plot(X_test[sorted_idx], Y_pred_dist[sorted_idx], color='red', label='Physics Model Fit')
plt.xlabel('RSSI (dBm)')
plt.ylabel('Distance (m)')
plt.title('Physics-Based Log-Distance Regression')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, r2_score

# 1. Predict distance using the physics formula on the test RSSI (X_test)
Y_pred_dist = predict_distance(X_test, p0, n_exponent)

# 2. Compare the predicted distances (Y_pred_dist) vs the ground truth (Y_test)
mae = mean_absolute_error(Y_test, Y_pred_dist)
r2 = r2_score(Y_test, Y_pred_dist)

print(f"Mean Absolute Error: {mae:.2f} meters")
print(f"R² Score: {r2:.2f}")

# 3. Optional: Validate in the Log-domain 
# (This tells you how well the 'line' fits before converting back to meters)
log_Y_pred = physics_model.predict(log_Y_test) # Predicted RSSI based on log distance
# Or vice versa depending on your model direction
```
