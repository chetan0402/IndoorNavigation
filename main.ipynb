```python
import pandas as pd
import numpy as np
```

```python
rows=[]
for key, values in data1.items():
    for val in values:
        rows.append((key,val))

for key, values in data2.items():
    for val in values:
        rows.append((key,val))
```

```python
df=pd.DataFrame(rows, columns=["RSSI","Distance"])
```

```python
df.to_csv("data.csv",index=False)
```

```python
data=df.to_numpy().astype(int)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.metrics import mean_absolute_error

X_train,X_test,Y_train,Y_test=train_test_split(data[:,0:1],data[:,1])

param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5, 10],
    'max_features': ['sqrt', 'log2', None]
}

grid_search = GridSearchCV(
    estimator=RandomForestRegressor(random_state=42),
    param_grid=param_grid,
    scoring='neg_mean_absolute_error', 
    cv=5,
    n_jobs=-1,
    verbose=2
)

grid_search.fit(X_train,Y_train)
model=grid_search.best_estimator_

predictions=model.predict(X_test)

print(np.percentile(np.abs(Y_test-predictions),95))
print(mean_absolute_error(Y_test,predictions))
```
